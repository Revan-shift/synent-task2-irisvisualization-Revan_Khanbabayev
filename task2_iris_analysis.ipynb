```python
# ==========================================
# Synent Technologies - Data Science Internship
# Task 2: Data Visualization (Iris Dataset)
# Author: [Your Name]
# ==========================================

# 1. Importing necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Loading the dataset
# Seaborn provides the Iris dataset internally
df = sns.load_dataset('iris')

# Displaying the first 5 rows to understand the structure
print("--- Dataset Preview ---")
print(df.head())

# Checking for basic information about the data
print("\n--- Dataset Info ---")
print(df.info())

# ------------------------------------------
# VISUALIZATION 1: Count Plot (Bar Chart)
# Goal: Check the distribution of flower species
# ------------------------------------------
# Updated version to avoid FutureWarning
plt.figure(figsize=(8, 5))
sns.set_style("whitegrid")

sns.countplot(x='species', data=df, hue='species', palette='viridis', legend=False)

plt.title('Distribution of Iris Species', fontsize=14)
plt.xlabel('Species', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.show()

# ------------------------------------------
# VISUALIZATION 2: Histogram
# Goal: Visualize the distribution of Petal Length
# ------------------------------------------
plt.figure(figsize=(8, 5))
sns.histplot(df['petal_length'], bins=20, kde=True, color='teal')
plt.title('Distribution of Petal Length', fontsize=14)
plt.xlabel('Petal Length (cm)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()

# ------------------------------------------
# VISUALIZATION 3: Scatter Plot
# Goal: Compare Sepal Length vs Sepal Width
# ------------------------------------------
plt.figure(figsize=(10, 6))
sns.scatterplot(x='sepal_length', y='sepal_width', hue='species', style='species', data=df, s=100)
plt.title('Sepal Length vs Sepal Width Comparison', fontsize=14)
plt.xlabel('Sepal Length (cm)', fontsize=12)
plt.ylabel('Sepal Width (cm)', fontsize=12)
plt.legend(title='Iris Species')
plt.show()

# ------------------------------------------
# VISUALIZATION 4: Pair Plot (Comprehensive Analysis)
# Goal: Visualize relationships between all features
# ------------------------------------------
pair_plot = sns.pairplot(df, hue='species', palette='husl', markers=["o", "s", "D"])
pair_plot.fig.suptitle('Pairwise Relationships of Iris Features', y=1.02, fontsize=16)
plt.show()

# ------------------------------------------
# FINAL SUMMARY: Statistical Insights
# ------------------------------------------
print("\n--- Summary Statistics ---")
print(df.describe())
```
